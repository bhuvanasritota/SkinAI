```python
def recommend_products(skin_type, concerns, top_k=5):

    user_skin = skin_mlb.transform([[skin_type]])

    user_concerns = concern_mlb.transform([concerns])

    user_features = np.hstack([
        user_skin,
        user_concerns
    ])

    similarity_scores = cosine_similarity(
        user_features,
        product_features
    )[0]

    recommendations = pd.DataFrame({
        'product': products_clean['product'],
        'similarity': similarity_scores
    })

    recommendations = recommendations.sort_values(
        by='similarity',
        ascending=False
    )

    return recommendations.head(top_k)
```

```python
def get_matched_concerns(product_concerns, user_concerns):
    product_concerns = product_concerns.split(',')

    return [
        concern
        for concern in user_concerns
        if concern in product_concerns
    ]
```

```python
def generate_reasons(skin_type, matched_concerns, user_skin_type):
    reasons = []

    if user_skin_type in skin_type.split(','):
        reasons.append(f"Matches your {user_skin_type} skin type")

    for concern in matched_concerns:
        reasons.append(f"Addresses your {concern} concern")

    return reasons
```

```python
def recommend_products_v2(skin_type, concerns, top_k=5):
    valid_concerns = set(concern_mlb.classes_)

    invalid_concerns = [
        concern for concern in concerns
        if concern not in valid_concerns
    ]

    if invalid_concerns:
        raise ValueError(
            f"Invalid concern(s): {invalid_concerns}"
        )

    # 1. Filter products by skin type
    eligible_products = products_clean[
        products_clean['skin_type'].str.contains(
            skin_type,
            case=False,
            na=False
        )
    ].copy()
    if eligible_products.empty:
        return pd.DataFrame(columns=[
        'product',
        'skin_type',
        'concern',
        'similarity',
        'concern_match',
        'final_score',
        'matched_concerns',
        'reasons'
    ])

    # 2. Get feature vectors of eligible products
    eligible_indices = eligible_products.index
    eligible_features = product_features[eligible_indices]

    # 3. Encode the user
    user_skin = skin_mlb.transform([[skin_type]])

    user_concerns = concern_mlb.transform([concerns])

    user_features = np.hstack([
        user_skin,
        user_concerns
    ])

    # 4. Calculate cosine similarity
    similarity_scores = cosine_similarity(
        user_features,
        eligible_features
    )[0]

    # 5. Calculate concern match count
    concern_match_count = eligible_products['concern'].apply(
        lambda x: sum(
            concern in x.split(',')
            for concern in concerns
        )
    )

    # 6. Convert concern matches to a score
    if concerns:
        concern_match_score = concern_match_count / len(concerns)
    else:
        concern_match_score = pd.Series(
        0.0,
        index=eligible_products.index
    )

    # 7. Calculate final score
    final_scores = (
        0.6 * similarity_scores
        + 0.4 * concern_match_score.values
    )

    # 8. Create recommendation DataFrame
    recommendations = eligible_products[
        ['product', 'skin_type', 'concern']
    ].copy()

    recommendations['similarity'] = similarity_scores
    recommendations['concern_match'] = concern_match_score.values
    recommendations['final_score'] = final_scores

    # 9. Find matched concerns
    recommendations['matched_concerns'] = (
        recommendations['concern'].apply(
            lambda x: get_matched_concerns(
                x,
                concerns
            )
        )
    )

    # 10. Generate explanations
    recommendations['reasons'] = recommendations.apply(
        lambda row: generate_reasons(
            row['skin_type'],
            row['matched_concerns'],
            skin_type
        ),
        axis=1
    )

    # 11. Sort by final score
    recommendations = recommendations.sort_values(
        by='final_score',
        ascending=False
    )

    # 12. Return Top-K
    return recommendations.head(top_k)
```
